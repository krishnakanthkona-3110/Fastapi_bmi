# What is Monolithic Architecture?

Monolithic architecture is a software design where the entire application is built as a single, unified unit.

# What is Fastapi?
FastAPI is a modern, high-performance Python web framework used to build APIs quickly and efficiently.

It is built on top of Starlette and Pydantic.

Starlette - Starlette manages how your API receives requests and sends back responses.

Pydantic - Pydantic is used to check if the data coming into your API is correct and in the right format.


# What is Latency?

Latency is how long it takes for data to travel from the client to the server and back.

Lower latency = faster response

Higher latency = slower response

Real-World Analogy 🚗

Distance between two cities = latency

Speed of the car = bandwidth

Even a fast car (high bandwidth) takes time if the distance (latency) is large.

2️⃣ WSGI (Web Server Gateway Interface)
What it is:

WSGI is a standard interface between Python web servers and web applications.

Used by:

Flask

Django

Django REST Framework

Characteristics:

✔ Synchronous (blocking)
✔ One request per worker
✔ Stable and simple

Example servers:

Gunicorn

In WSGI, each worker can process only one request at a time.
Multiple users can send requests concurrently as long as there are multiple workers available.
If all workers are busy, new requests must wait until a worker becomes free.

- WSGI = One request per worker, not one request per server
- Each worker handles one request at a time; concurrency depends on number of workers

# Process
A process is an isolated execution environment with its own memory, while a thread is a lightweight execution unit within a process that shares memory. In web servers, a worker can be implemented as either a process or a thread.

3️⃣ ASGI (Asynchronous Server Gateway Interface)
What it is:

ASGI is the modern successor to WSGI, designed for async and real-time applications.

Used by:

FastAPI

Django (async views)

Starlette

Characteristics:

✔ Async / await support
✔ Handles thousands of connections
✔ Supports WebSockets, background tasks

Example servers:

Uvicorn

# Uvicorn and Gunicorn are both Python web servers,

but they cater to different use cases. Uvicorn is a high-performance ASGI server, while Gunicorn is a mature WSGI server that also supports running Uvicorn as a worker.

For ASGI applications like those built with FastAPI. Uvicorn is generally preferred for its performance and asynchronous capabilities, while Gunicorn can be used for process management in production environments to handle multiple Uvicorn instances.

--- What is a Computed Field?

A field whose value is derived from other fields, not stored or passed in the request.

Example:

Input → price, quantity

Computed → total_price
