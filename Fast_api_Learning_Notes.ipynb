# Flask and FastAPI:

Flask and FastAPI are both Python web frameworks used to build web applications and REST APIs. Flask is a lightweight micro framework that provides only the core features and lets developers add functionality through extensions, making it simple and flexible.


FastAPI is a modern framework designed specifically for high-performance APIs. It supports asynchronous programming, automatic request validation with Pydantic, and generates interactive API documentation using Swagger UI and ReDoc.

In modern AI, machine learning, and microservice-based applications, FastAPI is often preferred because it handles concurrent requests efficiently and reduces boilerplate code.

# What is Fastapi?

FastAPI is a modern, high-performance Python web framework built on top of Starlette and Pydantic.

Starlette handles the core web layer — routing, requests, and responses, including async support —

while Pydantic handles automatic data validation, makes sure incoming and outgoing data matches the expected schema.

On top of that foundation, FastAPI adds things like automatic interactive API documentation via Swagger UI and ReDoc, and native support for asynchronous programming, which is especially useful for I/O-heavy backends — like calling an LLM provider or querying a vector database — where we want to handle multiple requests concurrently without blocking.

# Interview answer:

FastAPI is a modern, high-performance Python web framework for building structured APIs. It provides automatic request/response data validation through Pydantic, generates interactive API documentation using Swagger UI and ReDoc, and supports asynchronous programming for handling concurrent requests efficiently.

# Why you choose FastAPI:

I chose FastAPI mainly for three reasons. First, Pydantic-based validation means request/response schemas are enforced automatically, which catches malformed inputs early rather than failing deep inside business logic. Second, FastAPI's native async support is a real advantage for AI backends specifically — most of what these APIs do is I/O-bound: calling an LLM provider like Bedrock, querying a vector database, or hitting external tools — so being able to handle those concurrently without blocking the event loop meaningfully improves throughput under load. Third, the automatic OpenAPI/Swagger documentation just makes iteration faster — especially in a team setting, since other engineers or frontend consumers can see exactly what the API expects without needing separate documentation efforts. Compared to something like Flask, FastAPI gives you all of this out of the box instead of having to bolt on validation, docs, and async support separately.

2️⃣ WSGI (Web Server Gateway Interface)
What it is:

WSGI is a standard interface between Python web servers and web applications.

Used by:

Flask

Django

Django REST Framework

Characteristics:

✔ Synchronous (blocking)
✔ One request per worker
✔ Stable and simple

Example servers:

Gunicorn

In WSGI, each worker can process only one request at a time.
Multiple users can send requests concurrently as long as there are multiple workers available.
If all workers are busy, new requests must wait until a worker becomes free.

- WSGI = One request per worker, not one request per server
- Each worker handles one request at a time; concurrency depends on number of workers

3️⃣ ASGI (Asynchronous Server Gateway Interface)
What it is:

ASGI is the modern successor to WSGI, designed for async and real-time applications.

Used by:

Starlette

Characteristics:

✔ Async / await support
✔ Handles thousands of connections
✔ Supports WebSockets, background tasks

Example servers:

Uvicorn

# Uvicorn and Gunicorn are both Python web servers,

but they cater to different use cases. Uvicorn is a high-performance ASGI server, while Gunicorn is a mature WSGI server that also supports running Uvicorn as a worker.

For ASGI applications like those built with FastAPI. Uvicorn is generally preferred for its performance and asynchronous capabilities, while Gunicorn can be used for process management in production environments to handle multiple Uvicorn instances.
